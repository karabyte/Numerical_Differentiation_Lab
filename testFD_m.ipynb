```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming t, dfdtF, dfdtB, dfdtC, dfdtExact are defined elsewhere
# For demonstration, let's create some dummy data if not available
t = np.linspace(0, 2*np.pi, 100)
dfdtF = np.cos(t) + 0.1 * np.random.rand(100)
dfdtB = np.cos(t) - 0.1 * np.random.rand(100)
dfdtC = np.cos(t)
dfdtExact = np.cos(t)

plt.figure(figsize=(10, 6)) # Create a new figure

plt.plot(t, dfdtF, 'b', linewidth=1.2, label='Forward')
plt.plot(t, dfdtB, 'r', linewidth=1.2, label='Backward')
plt.plot(t, dfdtC, 'g', linewidth=1.2, label='Central')
plt.plot(t, dfdtExact, 'k--', linewidth=1.5, label='Exact')

plt.xlabel('t')
plt.ylabel('df/dt')

plt.legend()

plt.title('Numerical Differentiation of sin(t)')

plt.grid(True)
plt.show()
```
